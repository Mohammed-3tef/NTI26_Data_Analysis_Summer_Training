# NumPy Fundamentals: A Comprehensive Guide

## 1. Environment Setup & Version Check
Initialize NumPy and verify the installed runtime version.

```python
import numpy as np
print(f"NumPy Version: {np.__version__}")
```

## 2. Array Creation & Data Types
Demonstrating implicit type coercion in standard Python lists versus NumPy's homogeneous array requirement.

```python
# Python heterogeneous list
py_list = [1, 2, 2.5, "Ahmed"]
print("py_list:", py_list)

# Standard integer array
int_array = np.array([1, 2, 3])
print("int_array:", int_array)

# Coerced string array due to homogeneous type enforcement
str_array = np.array([1, 2, 3, "4"])
print("str_array:", str_array)
```

### Initializing Arrays from Scratch

```python
# Array of zeros (length 10)
zeros_arr = np.zeros(10)
print("zeros_arr:\n", zeros_arr)

# 3x5 array filled with floating-point ones
ones_arr = np.ones((3, 5))
print("ones_arr:\n", ones_arr)

# 3x5 array filled with a specific constant
full_arr = np.full((3, 5), 3.14)
print("full_arr:\n", full_arr)

# Linear sequence with step size (start, stop, step)
step_arr = np.arange(0, 21, 2)
print("step_arr:\n", step_arr)

# Evenly spaced sequence over a specified interval
linspace_arr = np.linspace(0, 1, 5)
print("linspace_arr:\n", linspace_arr)

# Uniformly distributed random floats between 0 and 1
rand_float_arr = np.random.random((3, 5))
print("rand_float_arr:\n", rand_float_arr)

# Random integers in a given range [0, 6)
rand_int_arr = np.random.randint(0, 6, (3, 5))
print("rand_int_arr:\n", rand_int_arr)

# 3x3 Identity matrix
identity_matrix = np.eye(3)
print("identity_matrix:\n", identity_matrix)

# Uninitialized array with unallocated memory values
empty_arr = np.empty((3, 5))
print("empty_arr:\n", empty_arr)
```

## 3. Array Attributes
Inspecting key dimensional and memory properties of a 2D NumPy array.

```python
# Sample 2D array
x = np.random.randint(0, 10, (3, 5))
print("x:\n", x)

# Display structural properties
print(f"Dimensions (ndim): {x.ndim}")
print(f"Shape:            {x.shape}")
print(f"Total elements:   {x.size}")
print(f"Data type:        {x.dtype}")
print(f"Item size (bytes):{x.itemsize}")
print(f"Total size (bytes):{x.nbytes}")
```

## 4. Array Indexing & Slicing
Accessing and modifying elements across 1D and 2D arrays.

```python
x1 = np.array([1, 2, 3, 4, 5])
print("x1:", x1)

# Element access
first_elem = x1[0]
print("first_elem:", first_elem)
last_elem = x1[-1]
print("last_elem:", last_elem)

# Slicing examples
sub_first_three = x1[0:3]
print("sub_first_three:", sub_first_three)
sub_every_other = x1[::2]
print("sub_every_other:", sub_every_other)
sub_reversed = x1[::-1]
print("sub_reversed:", sub_reversed)
```

```python
# Accessing elements in 2D array [row, column]
corner_val = x[-1, -1]
print("corner_val:", corner_val)

# Modifying elements
x[0, 0] = 10
print("Modified x:\n", x)

# Subsampling a 2x2 subarray
sub_2x2 = x[0:2, 0:2]
print("sub_2x2:\n", sub_2x2)

# Creating an explicit copy to avoid modifying the original array view
y = x[0:2, 0:2].copy()
y[0, 0] = 100
print("y (copy modified):\n", y)
```

## 5. Reshaping & Concatenation
Manipulating array dimensions and merging datasets.

```python
# Reshaping a 1D range into a 3x3 matrix
grid_3x3 = np.arange(0, 9).reshape((3, 3))
print("grid_3x3:\n", grid_3x3)

# Concatenating 1D arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
combined_1d = np.concatenate([arr1, arr2])
print("combined_1d:", combined_1d)

# Concatenating 2D arrays along axis 1 (columns)
grid = np.arange(1, 7).reshape((2, 3))
combined_2d = np.concatenate([grid, grid], axis=1)
print("combined_2d:\n", combined_2d)
```

## 6. Computation & Universal Functions (Ufuncs)
Vectorized operations for element-wise calculations.

```python
arr_vec = np.arange(4)
print("arr_vec:", arr_vec)

# Basic arithmetic operations
add_res = arr_vec + 5
print("add_res:", add_res)
sub_res = arr_vec - 5
print("sub_res:", sub_res)
mul_res = arr_vec * 2
print("mul_res:", mul_res)
div_res = arr_vec / 2
print("div_res:", div_res)
floor_div_res = arr_vec // 2
print("floor_div_res:", floor_div_res)
pow_res = arr_vec**2
print("pow_res:", pow_res)
mod_res = arr_vec % 2
print("mod_res:", mod_res)

# NumPy ufunc equivalents
np_add = np.add(arr_vec, 2)
print("np_add:", np_add)
np_sub = np.subtract(arr_vec, 2)
print("np_sub:", np_sub)
np_neg = np.negative(arr_vec)
print("np_neg:", np_neg)
np_mul = np.multiply(arr_vec, 5)
print("np_mul:", np_mul)
np_abs = np.absolute(-arr_vec)
print("np_abs:", np_abs)

# Trigonometric ufuncs
theta = np.linspace(0, np.pi, 3)
print("theta:", theta)
sin_vals = np.sin(theta)
print("sin_vals:", sin_vals)
cos_vals = np.cos(theta)
print("cos_vals:", cos_vals)
```

## 7. Aggregations & Missing Data Handling
Computing statistics and managing NaN values safely.

```python
# Vector reduction operations
arr_reduce = np.arange(4)
sum_reduced = np.add.reduce(arr_reduce)
print("sum_reduced:", sum_reduced)
prod_reduced = np.multiply.reduce(arr_reduce)
print("prod_reduced:", prod_reduced)

# Standard aggregations
min_val = np.min(arr_reduce)
print("min_val:", min_val)
max_val = np.max(arr_reduce)
print("max_val:", max_val)
total_sum = np.sum(arr_reduce)
print("total_sum:", total_sum)

# Multidimensional axis aggregations
column_sums = grid.sum(axis=0)
print("column_sums:", column_sums)

# Handling missing data (NaNs) safely
nan_array = np.array([1.0, 2.0, 3.0, 4.0, np.nan])
print("nan_array:", nan_array)

safe_sum = np.nansum(nan_array)
print("safe_sum:", safe_sum)
safe_mean = np.nanmean(nan_array)
print("safe_mean:", safe_mean)
safe_median = np.nanmedian(nan_array)
print("safe_median:", safe_median)
```

## 8. Boolean Masking & Sorting
Filtering values based on conditions and sorting arrays.

```python
data = np.array([1.0, 2.0, 3.0, 4.0, np.nan])
print("data:", data)

# Creating a boolean mask
mask = data > 2
print("mask:", mask)

# Applying the mask
filtered_data = data[data > 2]
print("filtered_data:", filtered_data)

# In-place sorting
unsorted_arr = np.array([50, 2, 25, 18, 7, 10])
unsorted_arr.sort()
print("sorted_arr:", unsorted_arr)

# Checking for NaN values
has_nan = np.isnan(np.nan)
print("has_nan:", has_nan)
```

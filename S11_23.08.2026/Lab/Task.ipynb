```python
import numpy as np
print("Numpy Version:", np.__version__)
```

```python
my_list = np.array([5, 7, 12, 52, 18, 8, 96, 2])
print("Data:", my_list)

my_list.sort()
print("Sorted Data:", my_list)

# Calculate Q1, Q2, Q3 and IQR
q1 = np.percentile(my_list, 25)
q2 = np.median(my_list)
q3 = np.percentile(my_list, 75)
iqr = q3 - q1

print(f"Q1: {q1}")
print(f"Q2: {q2}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
```

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

lower_outliers = my_list[my_list <= lower_bound]
upper_outliers = my_list[my_list >= upper_bound]

print("Lower Outliers:", lower_outliers)
print("Upper Outliers:", upper_outliers)
```

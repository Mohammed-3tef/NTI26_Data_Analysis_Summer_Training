## 2. Series & DataFrame Creation
A `Series` is a 1D indexed array; a `DataFrame` is a 2D table made of aligned Series sharing a common index.

```python
# A Series wraps values with an explicit index (unlike a plain NumPy array)
data = pd.Series([1, 2, 3, 4, 5])
print("data:\n", data)
print("data.values:", data.values)
print("data.index:", data.index)
```

```python
# Default integer index behaves like list/array indexing
print("data[0]:", data[0])
print("data[3]:", data[3])
```

### Explicit Indexing
A Series index can be any hashable labels (strings, dates, etc.), not just integers.

```python
# Custom, explicit index
data = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
print("data:\n", data)

print("data['a']:", data['a'])
print("data[['b', 'e']]:\n", data[['b', 'e']])
```

```python
# Series built directly from a dict: keys become the index, values become the data
dict_1 = {
    'Egypt': 300,
    'USA': 2000,
    'Canada': 1500
}

data = pd.Series(dict_1)
print("data:\n", data)

print("data['Egypt']:", data['Egypt'])

# Label-based slicing is INCLUSIVE of the stop label
print("data['Egypt':'Canada']:\n", data['Egypt':'Canada'])
```

### Building DataFrames

```python
# From a dict of Series/dicts sharing the same keys -- keys become columns, aligned on index
population = {'Egypt': 120, 'KSA': 80, 'USA': 300}
area = {'Egypt': 1, 'KSA': 1.8, 'USA': 5}

data = pd.DataFrame({'population': population, 'area': area})
print("data:\n", data)
```

```python
# From a list of dicts -- mismatched keys fill in with NaN
from_list_of_dicts = pd.DataFrame([{'a': 1, 'b': 2}, {'b': 5, 'c': 8}])
print("from_list_of_dicts:\n", from_list_of_dicts)
```

```python
# From a raw NumPy array, with explicit column names and index labels
from_numpy = pd.DataFrame(np.random.rand(3, 2), columns=['foo', 'bar'], index=list('abc'))
print("from_numpy:\n", from_numpy)
```

## 3. DataFrame Attributes
Inspecting the shape, labels, and dtypes of a realistic 2D DataFrame.

```python
# Sample dataset: US states, area and population
area = pd.Series({'California': 423967, 'Texas': 695662,
                   'New York': 141297, 'Florida': 170312,
                   'Illinois': 149995})

pop = pd.Series({'California': 38332521, 'Texas': 26448193,
                  'New York': 19651127, 'Florida': 19552860,
                  'Illinois': 12882135})

states = pd.DataFrame({'area': area, 'pop': pop})
print("states:\n", states)
```

```python
# Display structural properties
print(f"Shape:   {states.shape}")
print(f"Columns: {list(states.columns)}")
print(f"Index:   {list(states.index)}")
print(f"Dtypes:\n{states.dtypes}")
print(f"Values:\n{states.values}")
```

```python
# Column access via bracket notation returns a Series
print("states['area']:\n", states['area'])
print("states['pop']:\n", states['pop'])
```

```python
# Attribute-style access works ONLY when the column name doesn't collide with
# an existing DataFrame method -- 'pop' happens to be a built-in method name!
print("states.area is states['area']:", states.area is states['area'])
print("states.pop is states['pop']:", states.pop is states['pop'])
```

## 4. DataFrame Indexing & Slicing
Accessing and modifying elements with `loc` (label-based) and `iloc` (position-based) selection.

```python
# With an explicit, non-default integer index, plain bracket indexing becomes ambiguous
s = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
print("s:\n", s)

# Bracket INDEXING uses the explicit index (labels)
print("s[1]:", s[1])
print("s[3]:", s[3])

# Bracket SLICING, in contrast, falls back to the implicit positional index
print("s[1:3]:\n", s[1:3])
```

```python
# .loc is always label-based (and inclusive of the stop label)
print("s.loc[1:3]:\n", s.loc[1:3])

# .iloc is always position-based (and exclusive of the stop position), like plain Python
print("s.iloc[1:3]:\n", s.iloc[1:3])
```

```python
# Positional row/column slicing on a 2D DataFrame
print("states.iloc[:3, :2]:\n", states.iloc[:3, :2])

# Label-based row/column slicing (both ends inclusive)
print("states.loc[:'Illinois', :'pop']:\n", states.loc[:'Illinois', :'pop'])
```

```python
# Modifying a single element in place
states.iloc[0, 0] = 423967
print("modified states:\n", states)
```

```python
# .at / .iat are fast, single-scalar analogues of .loc / .iloc
print("states.at['Texas', 'pop']:", states.at['Texas', 'pop'])
print("states.iat[1, 1]:", states.iat[1, 1])
```

```python
# .copy() creates an explicit copy, so edits don't propagate back to the original
states_copy = states.copy()
states_copy.loc['Texas', 'pop'] = 0
print("states_copy:\n", states_copy)
print("original states (unchanged):\n", states)
```

## 5. Reshaping & Concatenation
Restructuring the index and merging separate DataFrames together.

```python
# Promoting a column to the row index with set_index()
orders_preview = pd.DataFrame({
    'OrderID': [1001, 1002, 1003],
    'Customer': ['Sara', 'Omar', 'Nour']
})
reindexed = orders_preview.set_index('OrderID')
print("reindexed:\n", reindexed)

# reset_index() moves the index back into an ordinary column
print("reindexed.reset_index():\n", reindexed.reset_index())
```

```python
# Concatenating two DataFrames by stacking rows
batch_1 = pd.DataFrame({'a': [1, 2], 'b': [3, 4]})
batch_2 = pd.DataFrame({'a': [5, 6], 'b': [7, 8]})

combined = pd.concat([batch_1, batch_2], ignore_index=True)
print("combined:\n", combined)
```

## 6. Computation & Vectorized Operations
Element-wise arithmetic between DataFrames/Series, automatically aligned on shared labels.

```python
print("states:\n", states)
```

```python
# Element-wise addition -- aligned automatically on the shared index
print("states + states:\n", states + states)
```

```python
# Subtracting a single-column DataFrame broadcasts by COLUMN alignment,
# so mismatched columns produce NaN
print("states - states[['pop']]:\n", states - states[['pop']])
```

```python
# .subtract(..., axis=0) instead broadcasts a Series down each ROW
print("states.subtract(states['pop'], axis=0):\n", states.subtract(states['pop'], axis=0))
```

```python
# Deriving a new column with a vectorized division -- no explicit loop needed
states['density'] = states['pop'] / states['area']
print("states:\n", states)
```

```python
# Boolean masking combined with fancy column selection via .loc
print("high density states:\n", states.loc[(states['density'] > 100), ['pop', 'density']])
```

## 7. Aggregations & Missing Data Handling
Computing summary statistics and managing NaN values safely.

```python
# Aggregate statistics over a column
print("total pop:", states['pop'].sum())
print("mean density:", states['density'].mean())
print("min area:", states['area'].min())
print("max area:", states['area'].max())
```

```python
print("states.info():")
states.info()
```

```python
print("states.describe():\n", states.describe())
print("states.head(3):\n", states.head(3))
print("states.tail(3):\n", states.tail(3))
```

```python
# isnull() flags missing entries; summing gives a per-column count
print("missing values per column:\n", states.isnull().sum())

# notnull() is the inverse mask -- keep only rows where EVERY value is present
print("states[states.notnull()]:\n", states[states.notnull()])
```

```python
# dropna(thresh=n) keeps rows with at least n non-null values
print("states.dropna(axis=0, thresh=3):\n", states.dropna(axis=0, thresh=3))
```

```python
# Introduce an actual missing value to see dropna in action
states.iloc[0, 2] = np.nan
print("states:\n", states)

# how='any' drops any row containing at least one NaN
print("states.dropna(how='any'):\n", states.dropna(how='any'))
```

## 8. Boolean Masking, Sorting & Data Cleaning
Filtering rows by condition, ordering values, and cleaning a realistic, messy dataset end to end.

```python
# A small sales-orders table to filter, sort, and clean
data = {
    "OrderID":   range(1001, 1011),
    "Customer":  ["Sara", "Omar", "Nour", "Sara", "Ali", "Mona", "Omar", "Sara", "Ali", "Nour"],
    "Region":    ["Cairo", "Giza", "Cairo", "Alexandria", "Giza", "Cairo", "Giza", "Cairo", "Alexandria", "Cairo"],
    "Category":  ["Electronics", "Furniture", "Electronics", "Office Supplies", "Furniture",
                  "Electronics", "Office Supplies", "Furniture", "Electronics", "Office Supplies"],
    "Product":   ["Laptop", "Chair", "Mouse", "Notebook", "Desk",
                  "Monitor", "Pen Set", "Chair", "Keyboard", "Notebook"],
    "Quantity":  [1, 2, 3, 10, 1, 2, 5, 1, 2, 8],
    "UnitPrice": [850, 120, 15, 2, 300, 220, 8, 120, 45, 2],
    "OrderDate": pd.to_datetime([
        "2024-01-05", "2024-01-07", "2024-01-15", "2024-02-01", "2024-02-10",
        "2024-02-14", "2024-03-02", "2024-03-05", "2024-03-20", "2024-03-25"
    ])
}

df = pd.DataFrame(data)
df['Revenue'] = df['Quantity'] * df['UnitPrice']
print("df:\n", df)
```

Real-world data usually starts life as a file rather than a Python literal --
`pd.read_csv()` / `pd.read_excel()` / `pd.read_json()` / `pd.read_sql()` load it straight into a DataFrame.

```python
print("region distribution:\n", df['Region'].value_counts(normalize=True))
print("unique customers:", df['Customer'].unique())
print("number of unique customers:", df['Customer'].nunique())
```

```python
# Boolean masking on a single condition
print("orders over 200 revenue:\n", df[df['Revenue'] > 200])
```

```python
# Combining multiple conditions with & (and) / | (or)
print("Cairo electronics orders:\n", df[(df['Region'] == 'Cairo') & (df['Category'] == 'Electronics')])

# .query() expresses the same filter as a readable string expression
print("same filter via .query():\n", df.query("Region == 'Cairo' and Category == 'Electronics'"))
```

```python
# In-place sorting by column, descending
df.sort_values('Revenue', ascending=False, inplace=True)
print("df sorted by Revenue:\n", df)
df.sort_index(inplace=True)
```

### Cleaning a Messy Copy
Deliberately introducing missing values and a duplicate row, then walking through the standard cleaning toolkit.

```python
dirty = df.copy()
dirty.loc[2, 'Quantity'] = np.nan
dirty.loc[5, 'Region'] = None
dirty = pd.concat([dirty, dirty.iloc[[0]]], ignore_index=True)
print("dirty:\n", dirty)
print("missing values per column:\n", dirty.isnull().sum())
```

```python
# dropna(subset=...) removes rows missing values in the given columns only
cleaned_dropped = dirty.dropna(subset=['Quantity', 'Region'])
print("cleaned_dropped:\n", cleaned_dropped)
```

```python
# fillna(): common strategies are
#   1) method='ffill' / 'bfill'  -> propagate neighboring values
#   2) value=scalar or {col: val} -> fill with a constant (or per-column constants)
filled = dirty.copy()
filled['Quantity'] = filled['Quantity'].fillna(filled['Quantity'].median())
filled['Region'] = filled['Region'].fillna('Unknown')
print("missing values after fillna:\n", filled.isnull().sum())
```

```python
# duplicated() flags rows that repeat an earlier row; drop_duplicates() removes them
print("duplicated rows:\n", filled.duplicated())
filled.drop_duplicates(inplace=True)
print("filled after dropping duplicates:\n", filled)
```

```python
# String accessor methods: .str.strip() (like SQL TRIM), .str.lower(), .str.title()
filled['Customer'] = filled['Customer'].str.strip().str.lower()
print("cleaned customer names:", filled['Customer'].unique())
```

```python
# astype() converts a column's dtype explicitly
print("dtypes before cast:\n", filled.dtypes)
filled['Revenue'] = filled['Revenue'].astype(float)
print("dtypes after cast:\n", filled.dtypes)
```

```python
# replace(old, new) swaps specific values
print("region with 'Unknown' relabeled:\n", filled['Region'].replace('Unknown', 'not specified'))

# clip(lower=, upper=) caps values into a range -- useful for taming outliers
print("clipped quantity:\n", filled['Quantity'].clip(lower=1, upper=10))
```
